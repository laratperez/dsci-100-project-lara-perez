#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# Creating a tidy dataset with the variables needed for the research question
ncaa_tidy <- ncaa |>
    select( # Keeping the 4 predictors, as well as identifying info
        year,
        school,
        seed,
        reg_wins_pct,
        conf_wins_pct,
        first_game_at_home,
        tourney_finish
    ) |>
    mutate(
        # Creating a column with binary factor outcome indicating wheter a team reached the top 16
        reached_top16 = as_factor(if_else(
                                    tourney_finish %in% c(
                                        "top_16_loss",
                                        "top_8_loss",
                                        "top_4_loss",
                                        "top_2_loss",
                                        "champ"
                                        ), "Yes", "No")
                    ),
        # Turning the column first_game_at_home to be a factor of Y or N
        first_game_at_home = as.factor(first_game_at_home),
        # Turning the column seed to be a factor since each number is a categorical label
        seed = as.factor(seed),
        # Renaming the outputs in first_game_at_home to be clearer and more readable
        first_game_at_home = fct_recode(first_game_at_home, 
                                        "Yes" = "Y", 
                                        "No" = "N")
    ) |>
    # Removing the original tournament finish column, since the new column created has the info needed for the research question
    select(-tourney_finish)

# Showing the first 6 rows of the tidied data
ncaa_tidy |> slice_head(n=6)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
ncaa_tidy |> 
summarize(
    # Applying the functions across all numeric columns
    across(
        where(is.numeric), 
        list(
            # Statistical summaries dropping missing values
            mean = ~mean(.x, na.rm = TRUE),
            min = ~min(.x, na.rm = TRUE),
            max = ~max(.x, na.rm = TRUE)
            ),
        # Setting the name for the new columns that will be created
        # Used "/" as a separator for easy organization later on
        .names="{.col}/{.fn}"
        )
    ) |>
# Moving the column names to be rows and statistical summaries to each be a column
# First, turning the variables and the respective data into columns
    pivot_longer(everything(), 
                 names_to = c("variable", "stat"), 
                 names_sep = "/") |>
# Then
    pivot_wider(names_from = stat, values_from = value)
```

```r
# Plot 1: Distribution of regional win percentages
plot_1 <- ncaa_tidy |> 
ggplot(aes(x=reg_wins_pct)) +
        # Setting bins to have steps of 5%
        geom_histogram(binwidth=5) +
        labs(x="Regional win percentage", 
             y="Number of teams", 
             title="Distribution of Regional win percentage across NCAA teams") +
        # Changing text size to be bigger and more readable
        theme(text = element_text(size = 14))

plot_1
```

```r
# Plot 2: Relative proportion of teams that did and did not reach the top 16
# Creating a df to find the relative proportion
home_proportions <- ncaa_tidy |>
    group_by(first_game_at_home) |>
    summarize(proportion = n()/nrow(ncaa_tidy)) |>
    select(first_game_at_home, proportion)
# home_proportions

# Plotting the relative proportion
plot_2 <- home_proportions |>
ggplot(aes(
        # Reversing the order of x-axis labels to have "Yes" then "No" for clarity
        x=fct_rev(first_game_at_home), 
        y=proportion)) +
    geom_bar(stat="identity") +
    labs(x="First game at home",
         y="Proportion of teams",
         title = "Proportion of Teams Playing Their First Game at Home") +
    theme(text = element_text(size = 14))

plot_2
```

**Visualization 1: Scatterplot of Regional vs. Conference Win Percentages**\
This scatterplot examines the relationship between regional and conference win percentages, two numerical predictors in the research question. This can show us if teams with higher conference win percentages generally also have higher regional win percentages. If the two predictors have a strong relationship, then that serves as an indication that they may provide overlapping information when predicting our outcome of top 16 achievement.

```r
# Creating a scatterplot to show the relationship between two numerical variables
visualization_1 <- ncaa_tidy |>
ggplot(aes(
    x = reg_wins_pct, 
    y = conf_wins_pct)) +
  # Reducing transparency to prevent overplotting
  geom_point(alpha=0.4) +
  labs(
    title = "Regional vs. Conference Win Percentages",
    x = "Regional win percentage",
    y = "Conference win percentage"
  ) +
  theme(text = element_text(size = 14))

visualization_1
```

**Visualization 2: Bar Chart of Top-16 Achievement Proportion by Tournament Seed**\
This visualization compares the proportion of teams reaching the top 16 across tournament seed categories. Since seed is categorical and the proportion of top-16 achievement is quantitative, a bar chart was used. Using a stacked bar chart allows us to compare the relative frequencies of the two outcomes within each seed. This helps identify differences in top-16 achievement across seeds, and assess whether the seed may be a useful predictor in the classification model.

```r
# Creating a bar graph to compare the values of categorical variable "seed"
visualization_2 <- ncaa_tidy |>
ggplot(aes(x=seed, 
           fill=reached_top16)
          ) +
  geom_bar(position="fill") +
  labs(
    x="Tournament seed",
    y="Proportion of teams",
    fill="Reached Top 16",
    title ="Proportion of Top-16 Achievement by Tournament Seed"
  ) +
    # Modifying y-axis to have a more detailed scale for eaier reading
  scale_y_continuous(breaks = seq(0, 1, by=0.2)) +
  theme(text = element_text(size = 14))

visualization_2
```
